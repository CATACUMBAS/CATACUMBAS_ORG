# Nomeie as cinco atividades fundamentais e explique o que a cascata e um método ágil fazem de diferente com elas.

A maior diferença está na frequencia, na ordem e na flexibilidade com que cada atividade é feita, enquanto o modelo cascata executa as atividades de forma linear, sequencial e rigida com fases isoladas, comunicação "essencial" apenas no começo, ou seja, um planejamento estático todo o projeto é definido apenas uma vez e posteriomente vem a construção do software

# Para cada cenário, escolha um modelo de processo e justifique ligando característica do problema a propriedade do modelo: (a) folha de pagamento regida por lei publicada; (b) painel exploratório para uma diretoria; (c) detector de fraude com técnica desconhecida pela equipe.

- Folha de pagamento regida por lei, pode ser mais viavel escolher o modelo cascata pois as leis em si ja foram publicadas, bem documentadas e dificilente mudarão durante o desenvolvimento, ou seja, previsibilidade com isso sendo mais viavel usar esse tipo de abordagem

- Painel exploratório para uma diretoria ja é um projeto mais "variavel" ja que o produto final nao vai mais depender de um documento "certo" mas das proprias pessoas que vao usar ele posteriormente, portanto um modelo de prototipagem seria o mais adequado na onde a diretoria explora e refina o que quer no projeto final

- Um detector de fraude com tecnica desconhecida pela equipe representa um risco arquitetual relativamente alto e alem disso tem o termo "desconhecido", um modelo de engenharia que faria bem para o projeto nesse sentido seria o espiral que foca na identificação e na mitigação de risco de cada ciclo.

#  Explique a diferença entre fatiar por valor e fatiar por risco, dando um exemplo de cada em um projeto de dados.


A diferença principal está no objetivo da segmentação: uma foca em gerar ganho financeiro rápido, enquanto a outra foca em mitigar incertezas técnicas ou de negócio.

Um exemplo usando fatiar por valor seria um sistema de recomendação de produto, ao inves de criar uma regra complexa o software pode fatiar a compra da pessoa em "pessoas que compram x tambem compram y" uma regra simples que aumenta o valor medio das compras

Um exemplo usando fatiar por risco é como nas perguntas acima um sistema de deteccção de fraude de cartão de credito, a primeira "fatia" foca estritamente na infraestrutura de dados: ingerir dados de transações em tempo real e aplicar uma regra estática

# Por que um protótipo deve ser combinado como descartável antes de ser construído? Descreva o que acontece quando isso não é feito.

Um protótipo deve ser combinado como descartável antes da construção para alinhar as expectativas da equipe e do cliente. Isso garante que todos saibam que aquele código ou design serve apenas para validar uma ideia, e não para virar o produto final.

# Um gestor afirma que a equipe deve «usar ágil para entregar mais rápido». Aponte o erro conceitual dessa frase e reescreva-a de forma tecnicamente correta

O erro conceitual da frase está em associar a agilidade diretamente ao aumento da velocidade de codificação ou escrita de software. Métodos ágeis não fazem uma equipe programar mais rápido, mas sim reduzir o desperdício. O foco do ágil é a entrega rápida de valor por meio de ciclos curtos (iterações), permitindo errar rápido, corrigir o rumo cedo e evitar a construção de recursos inúteis
