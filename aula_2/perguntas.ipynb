# Nomeie as cinco atividades fundamentais e explique o que a cascata e um método ágil fazem de diferente com elas.

A maior diferença está na frequencia, na ordem e na flexibilidade com que cada atividade é feita, enquanto o modelo cascata executa as atividades de forma linear, sequencial e rigida com fases isoladas, comunicação "essencial" apenas no começo, ou seja, um planejamento estático todo o projeto é definido apenas uma vez e posteriomente vem a construção do software

# Para cada cenário, escolha um modelo de processo e justifique ligando característica do problema a propriedade do modelo: (a) folha de pagamento regida por lei publicada; (b) painel exploratório para uma diretoria; (c) detector de fraude com técnica desconhecida pela equipe.

- Folha de pagamento regida por lei, pode ser mais viavel escolher o modelo cascata pois as leis em si ja foram publicadas, bem documentadas e dificilente mudarão durante o desenvolvimento, ou seja, previsibilidade com isso sendo mais viavel usar esse tipo de abordagem

- Painel exploratório para uma diretoria ja é um projeto mais "variavel" ja que o produto final nao vai mais depender de um documento "certo" mas das proprias pessoas que vao usar ele posteriormente, portanto um modelo de prototipagem seria o mais adequado na onde a diretoria explora e refina o que quer no projeto final

- Um detector de fraude com tecnica desconhecida pela equipe representa um risco arquitetual relativamente alto e alem disso tem o termo "desconhecido", um modelo de engenharia que faria bem para o projeto nesse sentido seria o espiral que foca na identificação e na mitigação de risco de cada ciclo.
