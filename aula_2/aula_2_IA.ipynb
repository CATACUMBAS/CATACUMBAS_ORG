# Escreva a descrição PEAS completa para: (a) um sistema de recomendação de produtos em um e-commerce; (b) um robô jogador de futebol; (c) um assistente de compras na internet. (Exercício 2.4 do livro.)

(a) Sistema de Recomendação de Produtos em um E-Commerce
Performance 
(Medida de Desempenho)
- Aumento na taxa de conversão de vendas.
- Aumento do tempo de retenção do usuário no site.
- Precisão e relevância das recomendações (cliques vs. rejeições).

Environment (Ambiente):
- Navegadores web e aplicativos mobile.
- Catálogo de produtos da loja.
- Base de dados de usuários e histórico de navegação.

Actuators (Atuadores):
- Seções de recomendação na interface ("Quem viu este produto também comprou").
- E-mails de marketing personalizados.
- Notificações push com ofertas sob medida.

Sensors (Sensores):
- Histórico de compras e buscas do cliente.
- Cliques, tempo de tela e interações do usuário.

(c) Assistente de Compras na Internet
Performance (Medida de Desempenho):
- Encontrar o menor preço disponível para o produto desejado.
- Garantir a confiabilidade e boa reputação da loja escolhida.
- Rapidez na execução e finalização da busca/compra.

Environment (Ambiente):
- Toda a internet pública (sites de e-commerce, buscadores, fóruns de promoção).
- Plataformas de pagamento e APIs de frete.

Actuators (Atuadores):
- Preenchimento automático de formulários e carrinhos de compras.Exibição de relatórios, 
- links e alertas na tela para o usuário.
- Envio de ordens de pagamento.

Sensors (Sensores):
- Comandos de texto ou voz fornecidos pelo usuário.
- Páginas HTML de produtos

# Defina com suas próprias palavras: agente, função de agente, programa de agente, racionalidade, autonomia. (Exercício 2.5 do livro.)

- Agente:  É qualquer entidade que consegue perceber o ambiente ao seu redor por meio de sensores e agir sobre ele
- Função do agente: É basicamente a logica usada para definir o comportamento dele
- Programa de Agente: é a implementação base para a função agente
- Racionalidade: é a caracteristica que define se o agente esta fazendo a coisa certa e que maximiza a sua chance de sucesso
- Autonomia: É a capacidade do agente de aprender e agir por conta própria

# Um agente racional pode tomar uma decisão correta e obter um péssimo resultado? E pode tomar uma decisão errada e obter um ótimo resultado? Justifique as duas respostas com a distinção entre desempenho esperado e desempenho real.

A justificativa para esse fenômeno reside estritamente na distinção entre desempenho esperado e desempenho real, agravada pelo fato de que a racionalidade depende do conhecimento disponível no momento da ação, enquanto o resultado final depende de fatores do ambiente que muitas vezes fogem ao controle ou à percepção do agente.
