1. Cinco passos do fluxo de contribuição do clone ao merge

*Clone do repositorio:* `git clone URL`  
*Artefato:* Cópia local do repositorio

*Criação da branch de trabalho*  
*Comando:*  
```git
git checkout -b feature
```  
*Artefato:* nova branch isolada para a mudança  

*Commits da Implementação*  
*Comando:*  
```git
git add .
git commit -m "mensagem"
```  
*Artefato:* commits contendo o código da funcionalidade ou correção.  

*Push da branch para o remoto*  
*Comando:*  
```git
git push origin feature
```  
*Artefato:* branch publicada no servidor (GitHub)  

*Pull request e merge*  
Abertura do PR -> Revisão -> Merge  
*Artefato:* Merge commit (ou squash merge) integrando a mudança na branch principal.

2. Falta 'quebra' do pull request por alteração ou seja cada uma das 38 alteracoes deveria ser um pull request e com uma mensagem bem descritiva da alteração feita, o risco de aprovar está justamente em problemas futuros que por exemplo seria necessario alguma mudança e o problema está relacionado a alguma mudança feita nesse pull request, como e a demora que esse programador demoraria para achar, revisar e dar a manutenção necessaria dentro dessas 38 alterações.

3. Bloqueante pois não trata o caso de lista vazia
Se a lista vier vazia, ou seja, ```items[0]``` causara erro
Sugestão: validar antes por exemplo ```if(!items.lenght) return;```

4. As 4 regras são  

1 Exigir PR para merge, isso evita que codigo sem revisão chegue na produção  
2 Exigir revisão por pares, garante que outra pessoa verifique a logica, levando segurança reduzindo falhas e melhora da qualidade  
3 Exigir Cl verde (testes passando), isso evita que código quebrado seja integrado, protegendo a estabilidade da branch principal  
4 Impedir force-push, evitando perda do histórco

5. Criar issue ```gh issue create```  
Abrir PR ```gh pr create```  
Trazer PR do colega para maquina ```gh pr chechout <numero>```  
Aprovar revisão ```gh pr review --approve```

6. Github flow: Simples, agil, muito usado para web e entrega continua, se sai bem para time pequenos/medios  
Git flow: Estruturado, muito usado para releases formais, projetos com ciclos de desenvolvimento longos  
Trunk-Based: Integração continua extrema, usado principalmente em times grandes, alta colaboração, por exemplo esse modelo é o usado em projetos como google ou microsoft

7 Revisão por pares tem uma má fam de ser reconhecido como um mecanismo para 'vigiar o desenvolvedor' quando ele realmente é um modo de foco para melhorar não só o produto quanto o proprio time, detecta erros cedo, compartilha conhecimento entre o time, padroniza o estilo e a pratica e etc.
