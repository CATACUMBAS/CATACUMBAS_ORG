1. No manifesto agil diz que certos valores 'são mais importantes que' outros mas não que o substituem, ou seja, do lado direito do manifesto continua tendo valor, apenas não deve prevalecer quando há conflitos.

2. Papéis:  
- Product Owner(PO) -> define a prioridade do produto  
- Scrum Master -> Facilita o processo, remove impedimentos  
- Developers -> Implementam o trabalho  

Artefatos:  
- Product Backlog -> Lista priorizada de tudo que pode ser feito  
- Sprint Backlog -> itens selecionados para a sprint + plano de execução  
- Incremento -> Entrega potencialemtne utilizavel ao final da sprint  

Cerimônias:  
- Sprint Planning -> planejar o que será feito  
- Dayli Scrum -> Sincronização diária  
- Sprint Review -> Demonstração do incremento  
- Sprint Tetrospective -> melhoria contínua do processo  
- Sprint -> ciclo de trabalho  

Quem decide quanto trabalho na sprint sao os proprios desenvolvedores, o PO pode priorizar mas não impoes quantidade sobre esses trabalhos

Um time construindo um produto novo com escopo negociavel a cada 2 semanas -> Scrum

XP inclui praticas como TDD, programação em par, refatoração e integração continua e se caso o time adota sprints curtas mas não adota XP ocorre problemas como Pressão para entregar rapidom codigo acumula dividas tecnicas e consequentemente o retrabalho aumenta, gerando pouca qualidade e etc

5. 
1 - Incerteza dos dados
Dados podem estar incompletos, sujos ou indisponíveis.
Ciclos curtos: validam cedo a qualidade dos dados.

2 - Incerteza do modelo
Não se sabe se o modelo estatístico/ML funcionará.
Ciclos curtos: experimentos rápidos reduzem risco.

3 - Incerteza do valor
Não está claro se a solução gerará impacto real.
Ciclos curtos: entregas frequentes permitem medir valor.

4 - Incerteza da integração
Dificuldade de colocar o modelo em produção.
Ciclos curtos: integração incremental revela problemas cedo.
