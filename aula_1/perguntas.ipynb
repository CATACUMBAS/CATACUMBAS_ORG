##  Liste três diferenças concretas entre um script que funciona e um software, usando um programa que você mesmo já escreveu como exemplo.

**Estrutura e complexidade**, um script é apenas um trecho unico de codigo que resolve um problema especifico, enquanto um software possui dezenas de milhares de arquivos, modulos, pacotes entre outras dependencias complexas alem de ele ser capaz de resolver inumeros problemas, um exemplo seria:

```python
# trecho de script
import pandas as pd

df = pd.DataFrame({
    "nome": ["Ana", "Bruno", "Carla", "Diego", "Elisa"],
    "idade": [23, 31, 19, 45, 28],
    "cidade": ["São Paulo", "Campinas", "São Carlos", "São Paulo", "Campinas"]
})

filtro_idade = df[df["idade"] > 25]
contagem = filtro_idade["cidade"].value_counts()
print(contagem)
```

Agora diferente de um **trecho de script**, um software é muito mais complexo do que apenas isso, contando com inumeros scripts, logica, organização sem contar que nao ha quebra de codigo, ou seja, ele mantem atvo mesmo que valores "errados" sejam passados, um exemplo está nesse pequeno projeto sobre finanças que está no meu github, conta com uma tela de login, um banco de dados, alem dos scripts para manter o codigo rodando [LINK](https://github.com/ArthurCost4/Projeto-Integrador-SENAC22.git)
